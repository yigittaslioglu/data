şeklinde olabilir bunu eksenlerde gösteriyor. kodun eksendeki halini yorumladı. bunlar temel karar kuralıdır. karar ağaçları heterojen veriyi karar kurallarına bölerek tahmin sonçları elde eder ve homojen hale getirir ve bunların ortalamasını alır, lightgbm, xgboost gibi karar ağacı yöntemleri en yüksek tahmin başarısına sahiptir. fakat görüntü işleme problemleri ya da metin işleme problemlerinde derin öğrenme temelli yöntemler ağaç yöntemlerine göre daha iyidir. 

bu kadar basitse bunu excelde yapalım neden uğraşıyoruz dediğimizde, evet yapabiliriz ancak soru şu, excelde bu x1 değişkenini neye göre bölmem gerektiğine nasıl karar verceğim. yani burada bir değere göre bölündü ancak bunu excelde elle bölmem gerekmektedir. biz normalde bir algoritmanın çalıştıktan sonra sonucunu görüyoruz burada. karar kuralları çıkarılmış. amaç burada karar kurallarını çıkarabilmek. karar kuralları çıkarıldıktan sonra zaten excelede uyarlanabilir. python ve sqle de uyarlanabilir. neden excel kullanmyoruz bunun algoritmik şekilde belirli referans noktalarına göre hareket ederek, kendimizi optimize ederek bu karar kurallarının ne olduğunu belirlemektir. burada mesela 1.7den böldük ama bunu 1.7den bölmeye neye göre karar verdik işte amaç bu kararın neye göre verildiğini bilebilmekte. bu alt gruplara neye göre ayırdık problemi var. ki2 gibi gini gibi entropi gibi rss(sse) gibi bazı metrikleri kullanarak bu karar kurallarına karar veriyor olacağız. dolayısıyla bunu excelde bu yüzden yapamayız. 

şimdi regresyon problemleri için cost fonksiyonumuz, diğer ifadesiyle loss  fonksiyonumuz, objective fonksiyonumuzu değerlendirelim, bu ağacı nasıl oluşturmalıyım sorusunun yanıtı rss diğer ifadesiyle sse değerinin minimum olduğu noktalardaki bölgelenmler, yapraklanmalar, kutulanmalardır. bu ne demektir. yi gerçek değerledir, yi - yrj burada r tahmin edilen değerlerde bölgeleri ifade etmektedir. regresyon problemleri için karar ağaçlarını kullandığımzda tahmin edilen değerler bölme işlemleri neticesinde ortaya çıkan kutulardan kalan gözlem birimlerinin ortalamasıdır. dolayısıyla bir karar noktasında bölme yaptık diyelimki dallanmalar yaptık bu bir kutudur, bu kutunun içerisinde kalan gerçek değerlerin ortalaması alınır, bunlar tahmin sonucu olarak değerlendirilir. dolayısıyla gerçek değerler ile tahmin edilen değerlerin farkı dediğimizde buradaki tahmin edilen değerler ilgili kutuların ortalama değerleridir. ben bu dallanmaları öyle bir derinleştirmeliyim ki toplam hatalar minimum olsun. kaç dal olacak, bir değişkeni nereden böleceğim, buna nasıl karar vereceğim, regresyon problemlerinde doğruyu nereye koyacağız problemi vardı en düşük hata veren noktaya koyuyorduk, burda da aynı durum vardır, bunu matematiksel bir optimizasyon problemine çeviriyorum, diyorum ki gerçek değerlerim ile tahmin edilen değerlerin hatasının en düşük olduğu noktada olsun bu yapraklar diyorum, yani değişkenlerin bölüm noktası.
