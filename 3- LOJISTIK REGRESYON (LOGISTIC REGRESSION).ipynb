### 7- LOG Loss

Şimdi bir optimizasyon gözlüğü takıyor olacağız, bu sefer optimize etmek istediğimiz bir fonksiyon var gözlüğüyle bakıp, fonksiyonumuzu optimize edeceğimiz fonksiyon olarak değerlendiriyor olacağız. Dikkat, log loss bir başarı metriğidir.

Log Loss = 1 / m((toplam sembolü m i=1) - yi * log(p(şapkalı yi))*log(1- p(şapkalı yi)))

bu başarı metriğini hem modelin başarısı için değerlendirebiliriz hem de optimizasyon yapmak üzere gerek hiperparametrelerde gerek ağırlıklarda, optimizasyon yapmak üzere ağırlıkları bulurken iterasyon sayısı açısından ya da diğer bazı olası farklı sınıflandırma algoritmalarında, makine öğrenmesi algoritmalarında, hiperparametreleri bulmak adına optimize etmek üzere odaklanacağımız fonksiyondur. peki yukarıda yazdıım fonksiyon nedir, cross entropi metriğidir. entropi ne demektir? entropi ne kadar düşükse o kaadr iyidir, ne kadar yüksekse o kadar kötüdür. entropi aslında çeşitlilik demektir. entropi ne kadar yüksekse o kadar çeşitliilk var demektir. entropi yüksekse o kadar bilgi vardır, çeşitlilik vardır, varyans vardır, ne kadar düşükse bunlar da o kadar azdır. örn: 1010 şeklinde bir sınıf olsun ve 1111 şeklinde sınfı olsun, ilk sınıfın entropisi daha yüksektir çünkü çeşitliliği fazladır. bu çeşitlilik konusunu yapmış olduğumuz tahminler ile elimizdeki gerçek değerler arasındaki durumu değerlendirmek için kullanacağız, bu formül cross entropi, binary (ikisli sınıflandırma)classification için kullanılan formu, bu formülde neler oluyor? bütün gözlem birimleri 1den m'ye kadar geziliyor, diyelim ki 1. gözlem birimindeyiz, yi yerine 1 koyuyorum ve formülün sağ tarafı gitti sadece sol taraf kaldı, logaritma ifadesi var ve bu gözlem birimi için 1 sınıfına ait olma olasılığı ifadesi var, bakıyorum tabloma 0.80 diyor, yani ne anlıyorum, -1 * log(0.80) şeklinde olmuş oluyor formülüm ve bu hesaplamayı yaptıktan sonra, bunu birinci gözlem birimim için yaptım, ikinci gözlem birimime geçiyorum, aynı işlem yalnızca 1 sınıfına ait olma olasılığı 0.48 olduğu için bunu yazdım 0.80 yerine, burda durup sonuçlarımı inceliyorum, ilk yaptığım hesaplamada sonuç daha küçük çıktı çünkü zaten 1e yakın bir sonuç entropisi daha küçük bir sonuç diğeri ise 0.48 arada bir değer, devam ediyorum, üçüncü gözlem birimim 0, 0 olduğu için bu sefer formlümün sol tarafı gitti sadece sağ tarafı kaldı, fyi gördüğüm yere bu sefer 0 yazıyorum, -1 * log(1- 0.30) yazıyorum bu sefer 0ın olma olasılığı değil 1 in olma olasılığını yazıyorum(ne gördüysem onu yazdım yine), çünkü zaten birden çıkaracağım için 0ın olma olasılığına erişeceğim. bu işlemleri devam ettiriyorum. burası kritik noktalardan biri, buradan elde edilecek olan log loss değerlerine göre optimizasyon işlemleri yapılıyor olacak, buradaki kritik nokta yine gerçek değerler ile yine tahmin edilen değerleri ölçme çabasıdır. bu ölçme çabasından dolayı ortalama alındığında yine ortalama hatamızı, kaybımızı, başarı ya da başarısızlığımızı ölçmüş olacağız. tüm yaptığımmız işlemler sonucu neticede bir log loss değeri gelir, sınıflandırma problemlerinde hem bir başarı değerlendirme metriği olarak kullanılır ne kadar küçük o kadar iyi şeklinde yorumlanır hem de optimize edilmek üzere diğer birçok görecek olduğumuz makine öğrenmesi yöntemlerinde de kullanacak olduğumuz objective foncitondur, amaç fonksiyonumuzdur, düşmesini beklediğimi - optimize ettiğimiz fonksiyonumuzdur.

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from sklearn.preprocessing import RobustScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score, confusion_matrix, classification_report, plot_roc_curve# başarı metriklerim
from sklearn.model_selection import train_test_split, cross_validate #holdout ve cross validationu aldım
```

```python
# bir değişkendeki outlier değerleri tanımlayabilmek için önce eşik değer hesabı
#yapılması gerekmektedir.bruada 75e 25 yerine 5e 95 kullanmayı tercih ettik bunun nedeni
#çok değişkenli etkiyi de göz ardı etmemek içindir ve aşırı aykırı olanlara dokunmak istememdir.

def outlier_thresholds(dataframe, col_name, q1=0.05, q3=0.95):
    quartile1 = dataframe[col_name].quantile(q1)
    quartile3 = dataframe[col_name].quantile(q3)
    interquantile_range = quartile3 -quartile1
    up_limit = quartile3 + 1.5 * interquantile_range
    low_limit = quartile1 - 1.5 * interquantile_range
    return low_limit, up_limit
```

```python
#Eşik değerleri hesapladıktan sonra bu değişkende aykırı değer var mı yok mu sorusu sormak için
#check_outlier fonksiyonumun olması gerekmektedir.
def chechk_outlier(dataframe, col_name):
    low_limit, up_limit = outlier_thresholds(dataframe, col_name) #burada önceki yazdığım fnksiyonu çağırtıyorum
    if dataframe[(dataframe[col_name] > up_limit) | (dataframe[col_name] < low_limit)].any(axis=None):
        return True 
    else:
        return False
```

```python
#yine eşik değerleri kullanarak eğer aykırı değer varsa bu aykırı değerleri silme bunları eşik değerler ile değiştir diyorum
def replace_with_thresholds(dataframe, variable):
    low_limit , up_limit = outlier_thresholds(dataframe, variable) #outlier thresholds bana limitleri versin
    dataframe.loc[(dataframe[variable] < low_limit), variable] = low_limit
    dataframe.loc[(dataframe[variable] > up_limit), variable] = up_limit
```

```python
def correct_bmi_value(value):
    """
    BMI sütunundaki tarih formatındaki değerleri gün.ay şeklinde float'a çevirir.

    Args:
        value: Düzeltilecek değer.

    Returns:
        Düzeltilmiş float değer veya hata durumunda None.
    """

    try:
        # DateTime'e dönüştürme işlemini daha spesifik hale getirme
        date_value = pd.to_datetime(value, errors='coerce', format='%Y-%m-%d')  # Varsayılan tarih formatı
        if pd.notna(date_value):
            # Gün ve ay bilgisini birleştirerek float'a çevir
            return float(f"{date_value.day}.{date_value.month}")

    except (ValueError, TypeError) as e:
        # Spesifik hata türlerini yakalama
        print(f"Hata: {e} - Değer: {value}")
        return None  # Hatalı değerleri None olarak ayır

    # Eğer datetime'e dönüştürme başarısız olursa, zaten bir sayıysa dönüştürmeye çalış
    try:
        return float(value)
    except ValueError:
        print(f"Hata: Değer {value} ne sayısal ne de tarih formatında.")
        return None

# BMI sütununu düzelt
df['BMI'] = df['BMI'].apply(correct_bmi_value)
```

# (10 - FOLD CROSS VALİDATİON)

```python
#Model doğrulama işlemlerine devam ediyorum, önceki bölümde train test diye ayırdık
#ancak hangi train test bunu bilemiyordum. öyle bir işlem yapmam lazım ki modelin doğrulama 
#sürecini en doğru şekilde ele almış olalım, bunu yapmanın yolu da çapraz doğrulama işlemidir,
#veri seti en başta 10 parçaya bölünür, 9 parçasıyla model kurulur bir parçasıyla test işlemi yapılır,
#daha sonra farklı bir 9 parçayla model kurulur, 1 parçayla test yapılır ve bu 10 katlı yapıldıktan sonra
#bütün test hatalarının, validasyon hatalarının ortalaması alınır. böylece model farklı veri setinin farklı parçalarıyla modellenip 
#farklı parçalarıyla test edilecek olduğundan aslında olası bazı senaryoları da barındırır 
#örn 1. parçada yüksek hata oranı elde ettik, ikinci de daha az gibi 10 parça için tekrarlanır ve daha sonra
#bu parçaların ortalaması alınır ve böylece zaten bir çok ihtimal değerlendirilmiş olur.

y = df["Outcome"] #bağımlı değişkenim
X = df.drop(["Outcome"], axis=1) #bağımsız değişkenlerim

log_model = LogisticRegression().fit(X, y) #burada bütün veriyi kullanarak bütün işlemi gerçekleştireceğiz
# burada tercih edilecek 2 şey olabilir, birincisi veri setimiz bolsa başta holdout yöntemiyle train tes olarak ikiye ayırıp train setine 10 katlı validasyon yapıp
# en son bir de test seti performansı incelenebilir, eğer veri setimiz bol değilse bütün veri  kullanılarak da işlem yapılabilir,
#bu durumda bütün veri kullanılarak da çapraz doğrulama işlemi gerçekleştirilebilir


#şimdi cross validasyonunu uygulayacağım, cross validasyo nder ki bana bir model ver, bağımlı ve bağımsız değişlenleri ver,
#kaç katlı cv yapacağını söyle onu ver ve en son olarak, ve hangi başarı metriklerini hesaplamak istediğimizi sorar

cv_results = cross_validate(log_model,
                           X, y,
                           cv=5,
                           scoring=["accuracy","precision","recall","f1","roc_auc"])
```

```python
#işlem tamamlandığı için test süreçlerine gideceğim
print(cv_results["test_accuracy"]) # test verilerim 5e böldüğüm için 5 oldu
print(cv_results["test_accuracy"].mean()) #ortalamasını aldığımdaysa nihai sonucumu elde etmiş olurum
```

```python
#diğer metriklerimi de getirip inceliyorum
print(cv_results["test_precision"]) # test verilerim 5e böldüğüm için 5 oldu
print(cv_results["test_precision"].mean())
```

```python
print(cv_results["test_recall"]) # test verilerim 5e böldüğüm için 5 oldu
print(cv_results["test_recall"].mean())
```

```python
print(cv_results["test_f1"]) # test verilerim 5e böldüğüm için 5 oldu
print(cv_results["test_f1"].mean())
```

```python
print(cv_results["test_roc_auc"]) # test verilerim 5e böldüğüm için 5 oldu
print(cv_results["test_roc_auc"].mean())
```
